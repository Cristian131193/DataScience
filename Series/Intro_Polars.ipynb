# Introducción a Polars: el futuro de Pandas

En este cuaderno aprenderás a usar Polars con ejemplos prácticos y funciones clave.

## Importar Polars

```
import polars as pl
```

## Cargar datos desde un archivo CSV

```
df = pl.read_csv('personas.csv')
df
```

## Crear un DataFrame desde un diccionario

```
datos = {
    'id': [1, 2, 3, 4, 5],
    'nombre': ['Ana', 'Luis', 'Sofía', 'Carlos', 'Marta'],
    'edad': [23, 35, 29, 42, 31],
    'ciudad': ['Buenos Aires', 'Córdoba', 'Rosario', 'Mendoza', 'La Plata'],
    'ingresos': [35000, 42000, 39000, 52000, 41000]
}
df2 = pl.DataFrame(datos)
df2
```

### 1. Ver las primeras filas: `head()`

```
df.head(3)
```

### 2. Ver la forma del DataFrame: `shape`

```
df.shape
```

### 3. Estadísticas descriptivas: `describe()`

```
df.describe()
```

### 4. Seleccionar columnas: `select()`

```
df.select(['nombre', 'edad'])
```

### 5. Filtrar filas: `filter()`

```
df.filter(pl.col('edad') > 30)
```

### 6. Crear o modificar columnas: `with_columns()`

```
df.with_columns((pl.col('ingresos') / 1000).alias('ingresos_miles'))
```

### 7. Ordenar datos: `sort()`

```
df.sort('edad', reverse=True)
```

### 8. Agrupar y resumir: `groupby()`

```
df.groupby('ciudad').agg(pl.col('ingresos').mean().alias('ingreso_promedio'))
```

### 9. Valores únicos: `unique()`

```
df.select('ciudad').unique()
```

### 10. Detección de nulos: `is_null()`

```
df.with_columns(pl.col('ingresos').is_null().alias('ingresos_nulo'))
```

### 11. Rellenar nulos: `fill_null()`

```
df.with_columns(pl.col('ingresos').fill_null(0))
```

### 12. Transformar a formato largo: `melt()`

```
df.melt(id_vars='nombre', value_vars=['edad', 'ingresos'])
```

### 13. Pivotear datos: `pivot()`

```
df.pivot(values='ingresos', index='ciudad', columns='nombre')
```

### 14. Unir DataFrames: `join()`

```
df_left = df.select(['id', 'nombre'])
df_right = df.select(['id', 'ciudad'])
df_left.join(df_right, on='id')
```

### 15. Aplicar funciones personalizadas: `map()`

```
df.with_columns(pl.col('nombre').map_elements(lambda x: x.upper()).alias('nombre_mayus'))
```

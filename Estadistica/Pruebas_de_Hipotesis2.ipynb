## Hipotesis Estadistica:

Es comprobar la veracidad de una hipotesis (o un supuesto) formulado sobre una poblacion en estudio.

### Contrastes de Hipotesis

Se encarga de contrastar la veracidad de las hipotesis estadistica (probar por medio de pruebas estadistica, si se cumple o no se cumple la hipotesis planteada).

### Tipos de Contrastes de Hipotesis

#### **Contrastes de conformidad:** Comprobar una hipotesis sobre los parametros de la poblacion. Ej: Validar si la nota media de un grupo de estudiantes es igual a 70.

#### **Contrastes de Homogeneidad:** Es comprobar 2 poblaciones con respecto alguno de sus paramtros. Ej: Quiero comprobar que el rendimiento de un grupo A es el mismo que el rendimiento del grupo B

#### **Contraste de Independencia:** Si existe relaciones en 2 poblaciones. Ej: contrastar si existe relacion en las notas de 2 poblaciones en materias diferentes.

**Contraste de Conformidad**

Formulacion de Hipotesis:

* H₀: La media de la poblacion es igual a 70 (μ = 70)
* H₁: La media de la poblacion no es igual a 70 (μ ≠ 70)



**Contraste de Homogeneidad**

Formulacion de Hipotesis:

* H₀: Las medias de las poblaciones son iguales (μ = μ₀)
* H₁: Las medias de las poblaciones no son iguales (μ ≠ μ₀)

**Contraste de Independencia**

Formulacion de Hipotesis:

* H₀: Las variables son independientes
* H₁: Las variables son dependientes

## Planteamiento de un Contraste de Hipotesis

Supone tomar una decision entre 2 hipotesis

#### **Hipotesis Nula:** Es la hipotesis conservadora. Lo que negamos falsedad H0
#### **Hipotesis Alternativa:** Es la negacion de la hipotesis Nula, osea es la afirmacion de lo que se pretende probar H1

## Prueba T Student

```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets

diabetes = datasets.load_diabetes()
df = pd.DataFrame(data=diabetes.data, columns=diabetes.feature_names)
df['target'] = diabetes.target

df
```

```python
# Supuestos para desnormalizar la edad
media_real_edad = 50  # Suponemos que la edad promedio real de la población es 50 años
desviacion_real_edad = 15  # Suponemos que la desviación estándar es 15 años

# Convertir la variable de edad (que está normalizada) a valores en años
df['edad_real'] = df['age'] * desviacion_real_edad + media_real_edad

# Revisamos la media y desviación estándar tras la transformación
print("Media de edad en la muestra:", df['edad_real'].mean())
print("Desviación estándar de la edad en la muestra:", df['edad_real'].std())

# Mostrar algunas filas de la nueva columna
df[['age', 'edad_real']].head()
```

Estableceremos un contexto general para formular hipótesis nula y alternativa. Supongamos que estamos interesados en investigar si la edad promedio de los pacientes diabéticos es mayor a 30 años. Entonces, las hipótesis serían:

Hipótesis nula (H0): La edad promedio de los pacientes diabéticos es menor igual a 30

*   Hipótesis nula (H0): La edad promedio de los pacientes diabéticos es menor igual a 30 años.
*   Hipótesis alternativa (H1): La edad promedio de los pacientes diabéticos es mayor a 30 años.

Dado que queremos investigar si la edad promedio es mayor a 30 años, realizaremos una prueba de hipótesis de una cola a la derecha.

Nivel de significacion = 0.05 para un IC = 95%

```python
# Parámetro de referencia para la prueba (por ejemplo, media poblacional de 50 años)
edad_promedio_poblacion = 50

# Prueba t para comparar la edad promedio de los pacientes con la población general
t_stat, p_valor = stats.ttest_1samp(df['edad_real'], edad_promedio_poblacion)

# Ajuste para una prueba de una cola a la derecha
if t_stat > 0:
    p_valor /= 2
else:
    p_valor = 1 - (p_valor / 2)

# Resultados
print("Estadístico t:", t_stat)
print("Valor p (una cola): {:.5f}".format(p_valor))
```

Si el valor p es menor a un nivel de significancia predefinido (por ejemplo, 0.05), entonces rechazamos la hipótesis nula y aceptamos la hipótesis alternativa.

Ahora, realizaremos una prueba de hipótesis de una cola a la izquierda. Supongamos que estamos interesados en investigar si el índice de masa corporal (IMC) promedio de los pacientes diabéticos es menor a 25.

* Hipótesis nula (H0): El IMC promedio de los pacientes diabéticos es mayor igual a 25.
* Hipótesis alternativa (H1): El IMC promedio de los pacientes diabéticos es menor a 25.

```python
# Prueba de hipótesis de una cola a la izquierda
imc_promedio_poblacion = 25
imc_pacientes = df['bmi']

t_stat, p_valor = stats.ttest_1samp(imc_pacientes, imc_promedio_poblacion)
p_valor /= 2 # Dividir entre 2, ya que queremos una prueba de una cola

if t_stat > 0:  # Si el estadístico t es positivo, el valor p para la prueba de una cola a la izquierda es 1 - p_valor
    p_valor = 1 - p_valor

print("Estadístico t:", t_stat)
print("Valor p: {:.3f}".format(p_valor))
```

Si el valor p es menor a un nivel de significancia predefinido (por ejemplo, 0.05), entonces rechazamos la hipótesis nula y aceptamos la hipótesis alternativa.

## Prueba ANOVA

**Constraste de Homogeneidad**

Escenario: Comparar la media del área de las células cancerosas en tres grupos de compacidad (bajo, medio, alto).

* Hipótesis nula (H0): Las medias del área de las células cancerosas son iguales en los tres grupos de compacidad.
* Hipótesis alternativa (H1): Al menos una de las medias del área de las células cancerosas es diferente en los tres grupos de compacidad.

```python
import pandas as pd
import numpy as np
import seaborn as sns
from scipy.stats import f_oneway
from sklearn.datasets import load_breast_cancer

# Cargar conjunto de datos
cancer = load_breast_cancer()
df = pd.DataFrame(data=cancer.data, columns=cancer.feature_names)

df.columns
```

```python
# Crear categorías para compacidad
bins = [0, 0.1, 0.2, np.inf]
labels = ['Bajo', 'Medio', 'Alto']
df['compactness_cat'] = pd.cut(df['mean compactness'], bins=bins, labels=labels)

# Separar el área de las células cancerosas en los tres grupos
area_low = df[df['compactness_cat'] == 'Bajo']['mean area']
area_medium = df[df['compactness_cat'] == 'Medio']['mean area']
area_high = df[df['compactness_cat'] == 'Alto']['mean area']

# Realizar prueba de ANOVA de un factor
f_stat, p_valor = f_oneway(area_low, area_medium, area_high)

print("Estadístico F:", f_stat)
print("Valor p:", p_valor)

#Prueba de Anova de 2 Factores sin Replicacion
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Crear la columna 'treatment'

median_texture = df['mean texture'].median()
df['treatment'] = np.where(df['mean texture'] > median_texture, 'Alta', 'Baja')

# Realizar ANOVA de dos factores
model = ols('Q("mean area") ~ compactness_cat * treatment', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)
```

```python
df
```

Interpretación: Si el valor p es menor a un nivel de significancia predefinido (por ejemplo, 0.05), rechazamos la hipótesis nula y concluimos que al menos una de las medias del área de las células cancerosas es diferente en los tres grupos de compacidad. Si el valor p es mayor, no podemos rechazar la hipótesis nula.

## Prueba Chi-Cuadrado

**Contraste de Conformidad**

Escenario: Verificar si hay una relación entre las especies de flores del conjunto de datos Iris y las categorías de longitud de pétalo (corto, mediano, largo).

* Hipótesis nula (H0): No hay una relación entre las especies de flores y las categorías de longitud de pétalo.
* Hipótesis alternativa (H1): Existe una relación entre las especies de flores y las categorías de longitud de pétalo.

```python
import pandas as pd
import numpy as np
import seaborn as sns
from scipy.stats import chi2_contingency
from sklearn.datasets import load_iris

# Cargar conjunto de datos
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target

# Crear categorías para la longitud del pétalo
bins = [0, 1.5, 3.5, 7]
labels = ['Corto', 'Mediano', 'Largo']
df['petal_length_cat'] = pd.cut(df['petal length (cm)'], bins=bins, labels=labels)

# Crear tabla de contingencia
contingency_table = pd.crosstab(df['species'], df['petal_length_cat'])

# Realizar prueba de Chi-Cuadrado
chi2, p_valor, dof, expected = chi2_contingency(contingency_table)

print("Estadístico Chi-Cuadrado:", chi2)
print("Valor p:", p_valor)
```

```python
contingency_table
```

Si el valor p es menor a un nivel de significancia predefinido (por ejemplo, 0.05), entonces existe una relación significativa entre la especie del Iris y la categorización de la longitud del pétalo.

La prueba de Chi-cuadrado que realizamos utilizando el conjunto de datos Iris para evaluar si la distribución de la longitud del pétalo en diferentes categorías (corto, mediano, largo) seguía una distribución esperada específica es un ejemplo de contraste de conformidad. En este caso, la hipótesis nula es que los datos siguen la distribución esperada y la hipótesis alternativa es que no lo hacen.

**Constraste de Independencia**

Escenario: Evaluar si la especie de Iris y la categoría de longitud del pétalo son independientes.

* Hipótesis nula: La especie de Iris y la categoría de longitud del pétalo son independientes.
* Hipótesis alternativa: La especie de Iris y la categoría de longitud del pétalo no son independientes.

```python
import pandas as pd
import numpy as np
import seaborn as sns
from scipy.stats import chi2_contingency
from sklearn.datasets import load_iris

# Cargar conjunto de datos
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target

# Crear categorías para la longitud del pétalo
bins = [0, 1.5, 3.5, 7]
labels = ['Corto', 'Mediano', 'Largo']
df['petal_length_cat'] = pd.cut(df['petal length (cm)'], bins=bins, labels=labels)

# Crear tabla de contingencia
contingency_table = pd.crosstab(df['species'], df['petal_length_cat'])

contingency_table.iloc[0:]


# Realizar prueba de Chi-Cuadrado de independencia
chi2, p_valor, dof, expected = chi2_contingency(contingency_table)

print("Estadístico Chi-Cuadrado:", chi2)
print("Valor p:", p_valor)
```

## Prueba de Kolmogorov-Smirnov

Escenario: Verificar si las distribuciones de las longitudes de pétalos de las especies Iris setosa e Iris versicolor son iguales.

* Hipótesis nula (H0): Las distribuciones de las longitudes de pétalos de las especies Iris setosa e Iris versicolor son iguales.
* Hipótesis alternativa (H1): Las distribuciones de las longitudes de pétalos de las especies Iris setosa e Iris versicolor son diferentes.

```python
import pandas as pd
import numpy as np
import seaborn as sns
from scipy.stats import ks_2samp
from sklearn.datasets import load_iris

# Cargar conjunto de datos
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target

# Separar las longitudes de pétalos de las especies Iris setosa e Iris versicolor
petal_length_setosa = df[df['species'] == 0]['petal length (cm)']
petal_length_versicolor = df[df['species'] == 1]['petal length (cm)']

# Realizar prueba de Kolmogorov-Smirnov
ks_stat, p_valor = ks_2samp(petal_length_setosa, petal_length_versicolor)

print("Estadístico KS:", ks_stat)
print("Valor p:", p_valor)
```

Si el valor p es menor a un nivel de significancia predefinido (por ejemplo, 0.05), entonces podemos concluir que las longitudes de los pétalos de las especies de Iris setosa e Iris versicolor no siguen la misma distribución subyacente.

## Prueba de Mann-Whitney U

Escenario: Comparar las áreas de los núcleos celulares entre tumores benignos y malignos en el conjunto de datos "Breast Cancer Wisconsin (Diagnostic)".

* Hipótesis nula (H0): Las áreas de los núcleos celulares tienen la misma distribución en ambos tipos de tumores.
* Hipótesis alternativa (H1): Las áreas de los núcleos celulares tienen distribuciones diferentes en ambos tipos de tumores.

```python
import pandas as pd
import numpy as np
from scipy.stats import mannwhitneyu
from sklearn.datasets import load_breast_cancer

# Cargar conjunto de datos
cancer = load_breast_cancer()
df = pd.DataFrame(data=cancer.data, columns=cancer.feature_names)
df['diagnosis'] = cancer.target

# Separar áreas de núcleos celulares en tumores benignos y malignos
areas_benign = df[df['diagnosis'] == 1]['mean area']
areas_malignant = df[df['diagnosis'] == 0]['mean area']

# Realizar prueba de Mann-Whitney U
u_stat, p_valor = mannwhitneyu(areas_benign, areas_malignant)

print("Estadístico U:", u_stat)
print("Valor p:", p_valor)
```

Interpretación: Si el valor p es menor a un nivel de significancia predefinido (por ejemplo, 0.05), rechazamos la hipótesis nula y concluimos que las áreas de los núcleos celulares tienen distribuciones diferentes en tumores benignos y malignos. Si el valor p es mayor, no podemos rechazar la hipótesis nula.

Escenario: Comparar las calificaciones de los estudiantes en dos exámenes.

* Hipótesis nula (H0): No hay diferencias significativas en las calificaciones de los estudiantes en los dos exámenes.

## Prueba de Kruskal-Wallis H

Escenario: Verificar si las distribuciones del contenido de alcohol en las tres clases de vino son iguales.

* Hipótesis nula (H0): Las distribuciones del contenido de alcohol en las tres clases de vino son iguales.
* Hipótesis alternativa (H1): Al menos una de las distribuciones del contenido de alcohol en las tres clases de vino es diferente.


```python
import pandas as pd
import numpy as np
from scipy.stats import kruskal
from sklearn.datasets import load_wine

# Cargar conjunto de datos
wine = load_wine()
df = pd.DataFrame(data=wine.data, columns=wine.feature_names)
df['wine_class'] = wine.target

# Separar el contenido de alcohol en las tres clases de vino
alcohol_class_0 = df[df['wine_class'] == 0]['alcohol']
alcohol_class_1 = df[df['wine_class'] == 1]['alcohol']
alcohol_class_2 = df[df['wine_class'] == 2]['alcohol']

# Realizar prueba de Kruskal-Wallis H
h_stat, p_valor = kruskal(alcohol_class_0, alcohol_class_1, alcohol_class_2)

print("Estadístico H:", h_stat)
print("Valor p:", p_valor)

```

Interpretación: Si el valor p es menor a un nivel de significancia predefinido (por ejemplo, 0.05), rechazamos la hipótesis nula y concluimos que al menos una de las distribuciones del contenido de alcohol en las tres clases de vino es diferente. Si el valor p es mayor, no podemos rechazar la hipótesis nula.

# Intervalos de Confianza

## Intervalo de confianza para la media:
Escenario: Estimar el intervalo de confianza de la media de la longitud del pétalo de la flor Iris setosa.

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
from sklearn.datasets import load_iris

# Cargar conjunto de datos
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target

# Filtrar las filas correspondientes a Iris setosa
setosa_petal_length = df[df['species'] == 0]['petal length (cm)']

# Calcular el intervalo de confianza de la media
mean, std = np.mean(setosa_petal_length), np.std(setosa_petal_length, ddof=1)
n = len(setosa_petal_length)
margin_of_error = stats.t.ppf(0.975, df=n-1) * std / np.sqrt(n)  # Nivel de confianza del 95%
lower, upper = mean - margin_of_error, mean + margin_of_error

print("Intervalo de confianza del 95% para la media:", (lower, upper))
```

Interpretación: Estamos 95% seguros de que la media de la longitud del pétalo de Iris setosa está entre los límites inferior y superior calculados.

## Intervalo de confianza para la proporción:
Escenario: Estimar el intervalo de confianza de la proporción de flores Iris setosa en el conjunto de datos Iris.

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
from sklearn.datasets import load_iris

# Cargar conjunto de datos
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target

# Calcular la proporción de flores Iris setosa
setosa_count = len(df[df['species'] == 0])
total_count = len(df)
proportion = setosa_count / total_count

# Calcular el intervalo de confianza de la proporción
margin_of_error = stats.norm.ppf(0.975) * np.sqrt(proportion * (1 - proportion) / total_count)  # Nivel de confianza del 95%
lower, upper = proportion - margin_of_error, proportion + margin_of_error

print("Intervalo de confianza del 95% para la proporción:", (lower, upper))
```

Interpretación: Estamos 95% seguros de que la proporción de flores Iris setosa en el conjunto de datos está entre los límites inferior y superior calculados.

## Intervalo de confianza para la diferencia de medias:
Escenario: Estimar el intervalo de confianza para la diferencia de medias entre las longitudes de pétalos de las flores Iris setosa e Iris versicolor.

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
from sklearn.datasets import load_iris

# Cargar conjunto de datos
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target

# Filtrar las filas correspondientes a Iris setosa e Iris versicolor
setosa_petal_length = df[df['species'] == 0]['petal length (cm)']
versicolor_petal_length = df[df['species'] == 1]['petal length (cm)']

# Calcular el intervalo de confianza para la diferencia de medias
mean_diff = np.mean(setosa_petal_length) - np.mean(versicolor_petal_length)
pooled_std = np.sqrt((np.var(setosa_petal_length, ddof=1) / len(setosa_petal_length)) + (np.var(versicolor_petal_length, ddof=1) / len(versicolor_petal_length)))
margin_of_error = stats.t.ppf(0.975, df=len(setosa_petal_length) + len(versicolor_petal_length) - 2) * pooled_std  # Nivel de confianza del 95%
lower, upper = mean_diff - margin_of_error, mean_diff + margin_of_error

print("Intervalo de confianza del 95% para la diferencia de medias:", (lower, upper))
```

Interpretación: Estamos 95% seguros de que la diferencia de medias entre las longitudes de pétalos de las flores Iris setosa e Iris versicolor está entre los límites inferior y superior calculados.

## Intervalo de confianza para la varianza:
Escenario: Estimar el intervalo de confianza para la varianza de la longitud del pétalo de la flor Iris setosa.

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
from sklearn.datasets import load_iris

# Cargar conjunto de datos
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target

# Filtrar las filas correspondientes a Iris setosa
setosa_petal_length = df[df['species'] == 0]['petal length (cm)']

# Calcular el intervalo de confianza para la varianza
n = len(setosa_petal_length)
sample_variance = np.var(setosa_petal_length, ddof=1)
lower = (n - 1) * sample_variance / stats.chi2.ppf(0.975, df=n-1)  # Nivel de confianza del 95%
upper = (n - 1) * sample_variance / stats.chi2.ppf(0.025, df=n-1)  # Nivel de confianza del 95%

print("Intervalo de confianza del 95% para la varianza:", (lower, upper))
```

Interpretación: Estamos 95% seguros de que la varianza de la longitud del pétalo de la flor Iris setosa está entre los límites inferior y superior calculados.
